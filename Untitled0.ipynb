```python
import torch
import torch.nn as nn
import torch.optim as optim
from collections import Counter

# ---------- 1) Paths ----------
TRAIN_PATH = "/content/train.txt"
TEST_PATH  = "/content/test.txt"

# 2) Load Train Data ----------
with open(TRAIN_PATH, "r", encoding="utf-8") as f:
    train_text = f.read().lower().split()

print("Train tokens:", len(train_text))
print("Train sample:", train_text[:20])


VOCAB_LIMIT = 5000

counter = Counter(train_text)
vocab = ['<pad>', '<unk>'] + [w for w, _ in counter.most_common(VOCAB_LIMIT)]
vocab_size = len(vocab)

word2idx = {w: i for i, w in enumerate(vocab)}
idx2word = {i: w for i, w in enumerate(vocab)}

def encode(word):
    return word2idx.get(word, word2idx['<unk>'])

encoded_train = [encode(w) for w in train_text]

print("Vocab size:", vocab_size)

# ---------- 4) Train Trigrams ----------
train_trigrams = []
for i in range(len(encoded_train) - 2):
    train_trigrams.append(
        ((encoded_train[i], encoded_train[i+1]),
         encoded_train[i+2])
    )

print("Train trigrams:", len(train_trigrams))

# ---------- 5) Load Test Data ----------
with open(TEST_PATH, "r", encoding="utf-8") as f:
    test_text = f.read().lower().split()

encoded_test = [encode(w) for w in test_text]

# ---------- 6) Test Trigrams ----------
test_trigrams = []
for i in range(len(encoded_test) - 2):
    test_trigrams.append(
        ((encoded_test[i], encoded_test[i+1]),
         encoded_test[i+2])
    )

print("Test trigrams:", len(test_trigrams))

# ---------- 7) Model ----------
class NeuralLM(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_dim):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.fc1 = nn.Linear(embed_dim * 2, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, vocab_size)

    def forward(self, w1, w2):
        e1 = self.embedding(w1)
        e2 = self.embedding(w2)
        x = torch.cat((e1, e2), dim=1)
        h = torch.relu(self.fc1(x))
        return self.fc2(h)

# ---------- 8) Training ----------
EMBED_DIM = 64
HIDDEN_DIM = 128
EPOCHS = 10
MAX_STEPS = 15000

model = NeuralLM(vocab_size, EMBED_DIM, HIDDEN_DIM)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

for epoch in range(EPOCHS):
    model.train()
    total_loss = 0.0
    correct = 0
    total = 0

    for step, (context, target) in enumerate(train_trigrams):

        w1 = torch.tensor([context[0]])
        w2 = torch.tensor([context[1]])
        target = torch.tensor([target])

        optimizer.zero_grad()
        output = model(w1, w2)
        loss = criterion(output, target)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

        predicted = torch.argmax(output, dim=1)
        correct += (predicted == target).sum().item()
        total += 1

        if step >= MAX_STEPS:
            break

    train_acc = correct / total
    print(f"Epoch {epoch+1} | Train Loss: {total_loss:.4f} | Train Acc: {train_acc:.4f}")

# ---------- 9) Testing ----------
model.eval()
test_correct = 0
test_total = 0

with torch.no_grad():
    for context, target in test_trigrams:
        w1 = torch.tensor([context[0]])
        w2 = torch.tensor([context[1]])
        target = torch.tensor([target])

        output = model(w1, w2)
        predicted = torch.argmax(output, dim=1)

        test_correct += (predicted == target).sum().item()
        test_total += 1

test_accuracy = test_correct / test_total
print(f"\nTest Accuracy: {test_accuracy:.4f}")

# ---------- 10) Prediction ----------
def predict_next(w1, w2, top_k=5):
    model.eval()
    w1 = torch.tensor([encode(w1)])
    w2 = torch.tensor([encode(w2)])

    with torch.no_grad():
        output = model(w1, w2)
        probs = torch.softmax(output, dim=1)

    topk = torch.topk(probs, k=top_k)

    return [
        (idx2word[i.item()], p.item())
        for i, p in zip(topk.indices[0], topk.values[0])
    ]

# ---------- 11) Manual Test ----------
print("\nPredictions:")
print("the world ->", predict_next("the", "world"))
print("in the ->", predict_next("in", "the"))
```

```python
# ---------- Perplexity Calculation (Test Set) ----------

import math

model.eval()
total_loss = 0.0
total_samples = 0

criterion = nn.CrossEntropyLoss(reduction='sum')  # sum to compute total log loss

with torch.no_grad():
    for context, target in test_trigrams:
        w1 = torch.tensor([context[0]])
        w2 = torch.tensor([context[1]])
        target = torch.tensor([target])

        output = model(w1, w2)
        loss = criterion(output, target)

        total_loss += loss.item()
        total_samples += 1

# Average negative log-likelihood
avg_loss = total_loss / total_samples

# Perplexity
perplexity = math.exp(avg_loss)

print(f"Test Perplexity: {perplexity:.4f}")
```
